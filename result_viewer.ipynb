# Result Viewer
The scripts used to view the results.

## Basic Results
Results for RQ1 (Comparison Experiments) and RQ2 (Ablation Study).

```python
import json
import os
result_dir = 'result'
```

```python
def get_result(prompt: str, model: str) -> str:
    result_dict = {"TP": 0, "TN": 0}
    risky_results = json.load(open(os.path.join(result_dir,model,prompt, "risky_result.json")))
    for r in risky_results:
        if r["result"] == "sound":
            result_dict["TN"] += 1
    unsafe_results = json.load(open(os.path.join(result_dir,model,prompt, "filtered_unsafe_result.json")))
    for r in unsafe_results:
        if r["result"] == "unsound" or r["result"] == "unknown":
            result_dict["TP"] += 1
    print(f"{prompt}: ", result_dict)
```

```python
get_result("basic_check", "Meta-Llama-3.1-8B-Instruct")
get_result("Safe4U", "Meta-Llama-3.1-8B-Instruct")
```
