# Result Viewer
The scripts used to view the results for RQ1 (Comparison Experiments) and RQ2 (Ablation Study).

```python
import json
import os
result_dir = 'result'
GT = {
    'T': 97,
    "F": 228
}
```

```python
def get_result(prompt: str, model: str) -> str:
    result_dict = {"TP": 0, "TN": 0}
    risky_results = json.load(open(os.path.join(result_dir,model,prompt, "risky_result.json")))
    for r in risky_results:
        if r["result"] == "sound":
            result_dict["TN"] += 1
    unsafe_results = json.load(open(os.path.join(result_dir,model,prompt, "filtered_unsafe_result.json")))
    for r in unsafe_results:
        if r["result"] == "unsound" or r["result"] == "unknown":
            result_dict["TP"] += 1
    result_dict["Accuracy"] = (result_dict["TP"] + result_dict["TN"]) / (GT["T"] + GT["F"])
    result_dict["Recall"] = result_dict["TP"] / GT["T"]
    result_dict["Precision"] = result_dict["TP"] / (result_dict["TP"] + (GT["F"] - result_dict["TN"]))
    result_dict["F1"] = 2 * result_dict["Precision"] * result_dict["Recall"] / (result_dict["Precision"] + result_dict["Recall"])
    print(f"{prompt}: ", result_dict)
```

```python
get_result("basic_check", "Qwen2-7B-Instruct")
get_result("Safe4U", "Qwen2-7B-Instruct")
```
